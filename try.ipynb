```python
import pandas as pd
import requests
import os
from time import sleep

# Setup
base_url = "http://pakirsa.gov.pk/Doc/"
output_dir = "irsa_data"
os.makedirs(output_dir, exist_ok=True)

dates = pd.date_range('2026-01-20', '2026-02-05')

# Track statistics
downloaded = 0
missing = 0
errors = 0

for i, date in enumerate(dates):
    url = f"{base_url}Data{date.strftime('%d-%m-%Y')}.pdf"
    filename = f"{output_dir}/Data{date.strftime('%Y-%m-%d')}.pdf"
    
    # Skip if already downloaded
    if os.path.exists(filename):
        print(f"[{i+1}/{len(dates)}] ✓ Already exists: {date.strftime('%Y-%m-%d')}")
        continue
    
    try:
        response = requests.get(url, timeout=10)
        
        if response.status_code == 200:
            with open(filename, 'wb') as f:
                f.write(response.content)
            downloaded += 1
            print(f"[{i+1}/{len(dates)}] ✅ Downloaded: {date.strftime('%Y-%m-%d')}")
        else:
            missing += 1
            print(f"[{i+1}/{len(dates)}] ❌ Not found: {date.strftime('%Y-%m-%d')} (Status: {response.status_code})")
    
    except Exception as e:
        errors += 1
        print(f"[{i+1}/{len(dates)}] ⚠️ Error: {date.strftime('%Y-%m-%d')} - {str(e)}")
    
    # Be polite to the server
    if (i + 1) % 10 == 0:
        sleep(1)

# Summary
print("\n" + "="*50)
print(f"✅ Downloaded: {downloaded}")
print(f"❌ Missing: {missing}")
print(f"⚠️ Errors: {errors}")
print(f"📁 Files saved to: {output_dir}/")
print("="*50)
```
